```python
keywords = [
    'literate',
    '10 or more years of schooling',
    'ever used the internet',
    'mobile phone',
    'electricity',
    'improved drinking-water',
    'improved sanitation',
    'clean fuel',
    'health insurance',
    'institutional births',
    'skilled health personnel',
    'fully vaccinated',
    'infant mortality',
    'under-five mortality',
    'stunted',
    'anaemic'
]

for keyword in keywords:
    print(f"\n--- {keyword} ---")
    for col in df_total.columns:
        if keyword.lower() in col.lower():
            print(col)
```

```python
women_internet = 'Women (age 15-49)  who have ever used the internet (%)'
men_internet = 'Men (age 15-49)  who have ever used the internet (%)'
women_mobile = 'Women (age 15-49 years) having a mobile phone that they themselves use (%)'

# Normalize each indicator using min-max scaling
df_project['Women Internet Score'] = (
    (df_project[women_internet] - df_project[women_internet].min()) /
    (df_project[women_internet].max() - df_project[women_internet].min())
) * 100

df_project['Men Internet Score'] = (
    (df_project[men_internet] - df_project[men_internet].min()) /
    (df_project[men_internet].max() - df_project[men_internet].min())
) * 100

df_project['Women Mobile Score'] = (
    (df_project[women_mobile] - df_project[women_mobile].min()) /
    (df_project[women_mobile].max() - df_project[women_mobile].min())
) * 100

# Overall Digital Inclusion Score
df_project['Digital Inclusion Score'] = (
    df_project['Women Internet Score'] +
    df_project['Men Internet Score'] +
    df_project['Women Mobile Score']
) / 3

print("Digital Inclusion Score created successfully!")
```

```python
female_literacy = 'Women (age 15-49) who are literate4 (%)'
male_literacy = 'Men (age 15-49) who are literate4 (%)'

female_schooling = 'Women (age 15-49)  with 10 or more years of schooling (%)'
male_schooling = 'Men (age 15-49)  with 10 or more years of schooling (%)'

education_columns = [
    female_literacy,
    male_literacy,
    female_schooling,
    male_schooling
]

# Min-max normalize each education indicator to 0–100
for col in education_columns:
    df_project[col + ' Score'] = (
        (df_project[col] - df_project[col].min()) /
        (df_project[col].max() - df_project[col].min())
    ) * 100

# Overall Education Score
df_project['Education Score'] = df_project[
    [col + ' Score' for col in education_columns]
].mean(axis=1)

print("Education Score created successfully!")
```

```python
electricity = 'Population living in households with electricity (%)'
water = 'Population living in households with an improved drinking-water source1 (%)'
sanitation = 'Population living in households that use an improved sanitation facility2 (%)'
clean_fuel = 'Households using clean fuel for cooking3 (%)'

infrastructure_columns = [
    electricity,
    water,
    sanitation,
    clean_fuel
]

# Normalize each indicator to a 0–100 scale
for col in infrastructure_columns:
    df_project[col + ' Score'] = (
        (df_project[col] - df_project[col].min()) /
        (df_project[col].max() - df_project[col].min())
    ) * 100

# Overall Infrastructure Score
df_project['Infrastructure Score'] = df_project[
    [col + ' Score' for col in infrastructure_columns]
].mean(axis=1)

print("Infrastructure Score created successfully!")
```

```python
health_positive = [
    'Households with any usual member covered under a health insurance/financing scheme (%)',
    'Institutional births (in the 5 years before the survey) (%)',
    'Children age 12-23 months fully vaccinated based on information from either vaccination card or mother\'s recall11 (%)'
]

health_negative = 'Infant mortality rate (per 1000 live births)'

# Normalize positive health indicators
for col in health_positive:
    df_project[col + ' Score'] = (
        (df_project[col] - df_project[col].min()) /
        (df_project[col].max() - df_project[col].min())
    ) * 100

# Reverse-score infant mortality because lower mortality is better
df_project['Infant Mortality Score'] = (
    (df_project[health_negative].max() - df_project[health_negative]) /
    (df_project[health_negative].max() - df_project[health_negative].min())
) * 100

# Calculate overall Health Score
health_score_columns = [
    col + ' Score' for col in health_positive
] + ['Infant Mortality Score']

df_project['Health Score'] = df_project[health_score_columns].mean(axis=1)

print("Health Score created successfully!")
```

```python
# Equal-weighted overall development score

df_project['Overall Development Score'] = (
    df_project['Education Score'] * 0.25 +
    df_project['Digital Inclusion Score'] * 0.25 +
    df_project['Infrastructure Score'] * 0.25 +
    df_project['Health Score'] * 0.25
)

print("Overall Development Score created successfully!")
```

```python
score_columns = [
    'Education Score',
    'Digital Inclusion Score',
    'Infrastructure Score',
    'Health Score'
]

# Find the strongest development dimension
df_project['Strongest Dimension'] = (
    df_project[score_columns]
    .idxmax(axis=1)
    .str.replace(' Score', '')
)

# Find the weakest development dimension
df_project['Weakest Dimension'] = (
    df_project[score_columns]
    .idxmin(axis=1)
    .str.replace(' Score', '')
)

print("Strength and gap analysis created successfully!")
```

```python
output = widgets.Output()

def update_dashboard(change):
    if change['name'] == 'value':
        with output:
            clear_output(wait=True)

            state_name = change['new']

            selected = analysis_df[
                analysis_df['States/UTs'] == state_name
            ].iloc[0]

            print(f"State/UT: {state_name}")
            print(f"Rank: {int(selected['Rank'])}")
            print(
                f"Overall Development Score: "
                f"{selected['Overall Development Score']:.1f}"
            )
            print(
                f"Strongest Dimension: "
                f"{selected['Strongest Dimension']}"
            )
            print(
                f"Weakest Dimension: "
                f"{selected['Weakest Dimension']}"
            )

            radar_data = pd.DataFrame({
                'Development Dimension': [
                    'Education',
                    'Digital Inclusion',
                    'Infrastructure',
                    'Health'
                ],
                'Score': [
                    selected['Education Score'],
                    selected['Digital Inclusion Score'],
                    selected['Infrastructure Score'],
                    selected['Health Score']
                ]
            })

            fig = px.line_polar(
                radar_data,
                r='Score',
                theta='Development Dimension',
                line_close=True,
                range_r=[0, 100],
                title=f'{state_name} — Development Profile'
            )

            fig.update_traces(fill='toself')

            fig.update_layout(height=600)

            fig.show()

state_dropdown.observe(
    update_dashboard,
    names='value'
)

display(output)
```

```python
def generate_state_insights(state_name):

    row = final_dashboard[
        final_dashboard['States/UTs'] == state_name
    ].iloc[0]

    overall_score = row['Overall Development Score']

    national_average = final_dashboard[
        'Overall Development Score'
    ].mean()

    difference_from_average = (
        overall_score - national_average
    )

    strongest = row['Strongest Dimension']
    weakest = row['Weakest Dimension']
    gap = row['Development Gap']

    print("=" * 65)
    print(f"DEVELOPMENT INSIGHTS — {state_name}")
    print("=" * 65)

    print(
        f"Overall Development Score: "
        f"{overall_score:.2f}"
    )

    print(
        f"National Average Score: "
        f"{national_average:.2f}"
    )

    if difference_from_average >= 0:
        print(
            f"Performance vs Average: "
            f"{difference_from_average:.2f} points above average"
        )
    else:
        print(
            f"Performance vs Average: "
            f"{abs(difference_from_average):.2f} points below average"
        )

    print(
        f"Strongest Dimension: {strongest}"
    )

    print(
        f"Weakest Dimension: {weakest}"
    )

    print(
        f"Development Gap: {gap:.2f} points"
    )
```

```python
def create_state_interpretation(state_name):

    row = final_dashboard[
        final_dashboard['States/UTs'] == state_name
    ].iloc[0]

    overall_score = row['Overall Development Score']

    national_average = final_dashboard[
        'Overall Development Score'
    ].mean()

    strongest = row['Strongest Dimension']
    weakest = row['Weakest Dimension']
    gap = row['Development Gap']

    difference = overall_score - national_average

    if difference > 0:
        comparison = (
            f"{abs(difference):.1f} points above the "
            f"average across the analyzed States/UTs"
        )
    elif difference < 0:
        comparison = (
            f"{abs(difference):.1f} points below the "
            f"average across the analyzed States/UTs"
        )
    else:
        comparison = "approximately equal to the average"

    interpretation = f"""
{state_name} has an Overall Development Score of
{overall_score:.1f}, which is {comparison}.

Its strongest dimension is {strongest}, while
{weakest} is its weakest dimension among the four
dimensions included in this analysis.

The difference between its strongest and weakest
dimension is {gap:.1f} points.
"""

    return interpretation
```
